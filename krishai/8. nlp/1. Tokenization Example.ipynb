```python
corpus = """
This function calculates the moving average of a numerical list using a specified window size. 
It is useful for smoothing time series data and identifying trends over a period. 
The result is returned as a list of averaged values, excluding incomplete windows at the beginning.
"""
```

```python
print(corpus)
```

```python
## Tokenization
###  Paragraphs --> Sentence
from nltk.tokenize import sent_tokenize

documents = sent_tokenize(corpus)
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization
###  Sentence --> Words
from nltk.tokenize import word_tokenize

word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize

wordpunct_tokenize(corpus)      # consider punctuations as separate token
```

```python
from nltk.tokenize import TreebankWordTokenizer

tokenizer=TreebankWordTokenizer()
tokenizer.tokenize(corpus)      # stop point isn't considered a new token
```
